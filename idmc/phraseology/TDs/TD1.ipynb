## Exercise 1

Collocations are defined as pairs of co-occurring words which are frequently appearing together. They tend to be considered as MWEs. We will explore different strategies in order to find collocations in text, without any additionnal information or resource.

- Open the file "TD1-files/pride_and_prejudices.txt" and convert it to a list of words
- Find and count the number of occurrence of each pair of co-occurring words
- Print the 10 most frequent pairs of words - what can you say about them ? are they collocations ?

```python
# ex1: open pride and prejudices and use the association measure to find frequent pairs of words
# first, show the most frequent collocations

with open("TD1-files/pride_and_prejudices.txt",'r',encoding='utf-8') as f:
    text = f.read()

word_list = text.replace("\n"," ").split(" ")
pairs_of_words = {}

for i in range(len(word_list)-1): # -1 to avoid out-of-range error
    
    word_pair = word_list[i] + " " + word_list[i+1]
    pairs_of_words.setdefault(word_pair,0)
    pairs_of_words[word_pair] += 1

most_frequent_pairs = sorted([(v,k) for k,v in pairs_of_words.items()],reverse=True)[:10]
print(most_frequent_pairs)
```

## Exercise 2

One way to find collocations is to compare each word in a pair of co-occurring word with the frequency of each word individually.
Check [Church and Hanks (1989)](https://aclanthology.org/P89-1010.pdf) for more information. During this exercise, we will apply their methodology.

- Fetch the number of occurrences of each word within Pride and Prejudices
- At the end of the last exercise, you should have a list or a dict containing each pair of words and its number of occurrences - make sure of that
- For each pair of word in this list or dict, compute its word association measure:
    - the calculus was explained during CM1 - check it if needed
    - you can use the **math** package to get the log2 of a number
- Print the 10 most frequent pairs of words - what can you say about them ? are they collocations ?
- Print the 10 least frequent pairs of words - what can you say about them ? are they collocations ?
- What is the big advantage of the word association measure then ?

```python
import math 
from collections import Counter

word_count = Counter(word_list)
results = {}

for k,v in pairs_of_words.items():

    word_1, word_2 = tuple(k.split(" "))
    
    prob_1 = word_count[word_1]/len(word_list)
    prob_2 = word_count[word_2]/len(word_list)
    prob_1_2 = v/len(word_list)

    association_measure = math.log2( (prob_1_2) / (prob_1*prob_2) )
    results.setdefault(k,association_measure)

most_frequent_pairs = sorted([(v,k) for k,v in results.items()],reverse=True)

for i in most_frequent_pairs[:10]:
    print(i[0],"\t",i[1])

print('\n...\n')

for i in most_frequent_pairs[-10:]:
    print(i[0],"\t",i[1])
```

```python
# now that we practiced a MWE discovery task, we are going to practice with a MWE identification task.
# show the number of time you find exact expressions in the sentences

import json

def open_json(path):
    "open a json file"
    with open(path,'r',encoding='utf-8') as f:
        data = json.load(f)
    return data

sentences = open_json("TD1-files/sentences.json")
expressions = open_json("TD1-files/expressions.json")

token_matches = 0
lemma_matches = 0

for sentence in sentences:
    sent_token = " ".join(sentence["token"])
    sent_lemma = " ".join(sentence["lemma"])

    for mwe in expressions:
        
        if mwe in sent_token:
            token_matches += 1

        if mwe in sent_lemma:
            lemma_matches += 1

print(token_matches)
print(lemma_matches)
```
