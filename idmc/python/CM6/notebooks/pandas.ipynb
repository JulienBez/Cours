```python
# dataset: https://github.com/mwaskom/seaborn-data/blob/master/exercise.csv

import pandas as pd

# Open a file

data = pd.read_csv(
    "exercise.csv",
    sep=",",
    header="infer",
    index_col=0,
    encoding="utf-8"
)

# In jupyter notebook, dataframes are prettier without using print()
data
```

```python
# Access elements by id

data.head(1)
data.tail(1) 

data.loc[0]
data.loc[0:10]
```

```python
# Looping over rows

# Will only show the column names
for i in data:
    print(i)

# Correct loop over rows with iterrows()
for index,row in data.iterrows():
    print(row["id"],row["kind"]) # you can also just do "print(row)"
```

```python
# Looping over columns

data.columns # show the columns
data.columns = ["id","diet","pulse","time","kind"] # change the column names

for i in data["pulse"]:
    print(i)
    
for i,pulse in enumerate(data["pulse"]):
    print(pulse,data["kind"][i])
```

```python
# Add a row

data = pd.concat(
    [
        pd.DataFrame([[1,"test",0,"15 min","test"]], columns=data.columns),
        data
    ],
    ignore_index=True
)

data
```

```python
# Add a column

new_column = []

for i in range(len(data)):
    new_column.append(f"col number {i}")
    
data["description"] = new_column

data
```

```python
# Modify a cell value

data.loc[0,"description"] = "ITS A TRAP"

data.head()
```

```python
# Modify every cell value in a column

for i,value in enumerate(data["description"]):
    data.loc[i,"description"] = value.replace("number","")
    
data
```

```python
# Delete a row

data = data.drop(0)
data = data.drop([1,2,3])

data
```

```python
# Delete a column

data = data.drop(columns=["kind"])

data
```

```python
# Save a file

data.to_csv(
    "exercise_v2.csv",
    sep=",",
    header=True,
    index=True,
    encoding="utf-8"
)
```
