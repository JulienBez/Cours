```python
# Based on the work of Aryan Kumar Mishra
# https://medium.com/@mishraavyan3/iris-dataset-classification-using-python-scikit-learn-b6ecd724211f

from sklearn.datasets import load_iris

iris = load_iris()

print(iris.data)
print(iris.target)
```

```python
# To better see the dataset, lets convert it to a pandas dataframe
# We wont use this pandas dataframe after

import pandas as pd

df = pd.DataFrame(iris.data, columns=iris.feature_names)

df
```

```python
# We want to do a classification task on the iris dataset

# Values within the dataset (for each flower)
#    - Sepal length
#    - Sepal width
#    - Petal length
#    - Petal width

# Possible labels (for each flower)
#    - 0 => Setose
#    - 1 => Versicolor
#    - 2 => Virginica

# Goal: determine if a flower is a setose, a versicolor or a virginica

# To do that, we need to train a Machine Learning Model
#   - we need a train partition, to train a statistical model
#   - we need a test partition, to test our model and compare its result to a gold standard

# In the following cells, we are going to do that with scikit-learn
```

```python
# Step 1: split features and labels

x = iris.data   # sepal and petal length + width
y = iris.target # label, either 0 or 1 or 2

print(x)
print(y)
```

```python
# Step 2: split the dataset in two partitions - train and test

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42
)

# test_size is the % of the total dataset in test
# random_state is optional but help for reproductibility => same value = same result

print(x_test)
print(y_test)
```

```python
# Step 3: initialize a Machine Learning model

from sklearn.linear_model import LinearRegression

model = LinearRegression()    # we initialize the model
model.fit(x_train, y_train)   # we give training data to the model

model.score(x_train, y_train) # we check the accuracy of our model on the train partition 
```

```python
# Step 4: test the model on data it hasn't seen before

print(model.predict(x_test))      # show model prediction for each element in test partition
print(model.score(x_test,y_test)) # show model accuracy on the test partition
```

```python
# Bonus step: a good practice when doing Machine Learning

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression

model = LinearRegression()

scores = cross_val_score( # what do we mean by cross-validation ?
    model,
    x,
    y,
    cv=5                  # what does this number correspond to ?
)

print(scores)
print(scores.mean())
```

```python
# ONE MORE EXERCISE - similarities with scikit-learn

# We can use sklearn to compute the similarity between two strings

from sklearn.metrics import pairwise_kernels
from sklearn.feature_extraction.text import CountVectorizer

sentence_A = "Im a cool sentence, look at me!"
sentence_B = "Im a VERY cool sentence, look at me instead!"
sentence_C = "Im a sweet potatoe"

sentence_list = [sentence_A,sentence_B,sentence_C]

vectorizer = CountVectorizer()              # we initiate a vectorizer, what does it do ?
X = vectorizer.fit_transform(sentence_list) # we apply the vectorizer on our sentences 

# The following line does something. Can you explain it ?

sim_matrice = pairwise_kernels(X,metric="cosine")
print(sim_matrice)
```
