```python
# dataset: https://www.kaggle.com/datasets/venky73/spam-mails-dataset?resource=download

# Before all, we load the dataset we want to use

import pandas as pd

data = pd.read_csv(
    "spams.csv",
    sep=",",
    header="infer",
    index_col=0,
    encoding="utf-8"
)

data
```

```python
# First exercise: find similar looking mails

# Step 1: transform the mails to vectors and compute a cosine similarity on them
from sklearn.metrics import pairwise_kernels
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(data["text"])

sim_matrice = pairwise_kernels(X,metric="cosine")

print(sim_matrice)
```

```python
for sim_row in sim_matrice:
    similar = [i for i in sim_row if i > 0.5]
    print(len(similar)-1) # why -1 ?
```
